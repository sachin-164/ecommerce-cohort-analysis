```python
import pandas as pd

df = pd.read_csv('online_retail.csv', encoding = 'unicode_escape')
df.info()
```

```python
df.head(10)
```

```python
df.columns
```

```python
# Drop rows without a CustomerID
df = df.dropna(subset=['CustomerID'])
```

```python
# Convert InvoiceDate from Text to real DateTime object
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df.head(100)
```

```python
# Filter out returns and cancellations
df = df[df['Quantity'] > 0]
```

```python
# converting CustomerID from float to integer 
df['CustomerID'] = df['CustomerID'].astype(int)
```

```python
# num of duplicates exist
df.duplicated().sum()
```

```python
# Drop duplicate rows
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# Check the dataset again
df.info()
```

```python
print(df.columns)
```

# Feature Engineering

```python
df['InvoiceMonth'] = df['InvoiceDate'].dt.to_period('M')
```
